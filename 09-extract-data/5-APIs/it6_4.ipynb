```python
import requests
import pandas as pd
import time
pd.set_option('display.max_rows', None)
def get_entreprises(siren):
    #siren = "892728080" 
    url = "https://api.insee.fr/api-sirene/3.11/siren/"
    headers = {
        "X-INSEE-Api-Key-Integration": "ad241078-6180-4d8b-a410-7861805d8b43",
        "accept": "application/json",
    }
    params  = {
        'siren' : siren,
    }
    r = requests.get(url, headers=headers, params=params)
    print(f"Temps de réponse: {r.elapsed.total_seconds()} secondes")
    if r.status_code == 200:
        data = r.json()#["uniteLegale"]
        if not data:
            return None
        df = pd.json_normalize(data)
        return df
    else:
        print(r.status_code , " : PAGE NOT FOUND POUR ENTREPRISE :",siren)
        return None

```

```python
df_csv = pd.read_csv("entreprises.csv")
resultat = pd.DataFrame()
cpt = 0

for siren in df_csv['siren']:
    
    df_temp = get_entreprises(siren)
    if df_temp is not None and not df_temp.empty:
        resultat = pd.concat([resultat,df_temp])
    time.sleep(2)
    cpt +=1
    if cpt > 50:
        break
        
resultat.to_csv("entreprises_insee.csv", index=False)
```
