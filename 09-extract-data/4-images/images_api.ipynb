```python
import requests
import pandas as pd
import json


BASE_URL = "https://picsum.photos/v2/list"
params = {
    "limit" : 50
}
response = requests.get(BASE_URL, params=params)
```

```python
if response.status_code == 200:
    liste = response.json()
else:
    print("page not found")
```

```python
df = pd.json_normalize(liste)
df = df.set_index("id")
authors = df["author"].unique()
authors_r = [x for x in authors if 'r' in x]
authors_r
```

```python
urls_images = df['download_url'].where(df['author'].isin(authors_r)).dropna()
urls_images
```

```python
urls_images.to_json("images.json", orient="records", indent=4)
```

```python
images = urls_images.sample(10)
cpt = 1
for link in images:
    image = requests.get(str(link)+".jpg")
    with open(f"photo{cpt}.jpg", "wb") as f:
        f.write(image.content)
    cpt +=1
```
