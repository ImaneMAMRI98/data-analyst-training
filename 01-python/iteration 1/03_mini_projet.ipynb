### `est_assez_long`

```python
def est_assez_long(pwd: str, min_len: int = 8) -> bool:
    """True si le mot de passe a au moins `min_len` caractères."""
    # TODO
    if len(pwd)<min_len:
        return False
    else:
        return True
    raise NotImplementedError
```

### `contient_majuscule`

```python
def contient_majuscule(pwd: str) -> bool:
    """True si le mot de passe contient au moins une lettre A-Z."""
    # TODO
    return any(char.isupper() for char in pwd)
    raise NotImplementedError
```

### `contient_chiffre`

```python
def contient_chiffre(pwd: str) -> bool:
    """True si le mot de passe contient au moins un chiffre 0-9."""
    # TODO
    return any(char.isdigit() for char in pwd)
    raise NotImplementedError
```

### `contient_caractere_special`

```python
def contient_caractere_special(pwd: str) -> bool:
    """True si le mot de passe contient au moins un caractère de CARACTERES_SPECIAUX."""
    # TODO
    for char in pwd:
        if char in CARACTERES_SPECIAUX:
            return True
    return False
    raise NotImplementedError
```

### `score_force`

Retourne un entier de **0 à 4** : nombre de règles satisfaites parmi les 4 ci-dessus
(longueur, majuscule, chiffre, spécial).

C'est juste de la composition des 4 fonctions précédentes.

```python
def score_force(pwd: str) -> int:
    """Retourne un entier de 0 à 4 : nombre de règles respectées
    parmi (longueur, majuscule, chiffre, caractère spécial).
    """
    # TODO
    res = 0
    if est_assez_long(pwd):
        res +=1
    if contient_majuscule(pwd):
        res +=1    
    if contient_chiffre(pwd):
        res +=1    
    if contient_caractere_special(pwd):
        res +=1
    return res
    raise NotImplementedError
```

### `temps_cassage_estime`

Retourne le temps moyen en **secondes**, en suivant la formule ci-dessus.

- Détermine la taille de l'alphabet en fonction des classes présentes.
- Si le mot de passe est vide, retourne `0.0`.

```python
def temps_cassage_estime(pwd: str, tentatives_par_seconde: float = 1e9) -> float:
    """Estime le temps moyen (en secondes) pour casser le mot de passe par force brute.

    - taille_alphabet = somme des classes utilisées (minuscules=26, majuscules=26,
      chiffres=10, spéciaux=32).
    - combinaisons = taille_alphabet ** len(pwd)
    - tentatives moyennes = combinaisons / 2
    - temps = tentatives_moyennes / tentatives_par_seconde

    Retourne 0.0 si le mot de passe est vide.
    """
    # TODO
    if pwd == "":
        return 0.0
    taille_alphabet = 0
    if any(char.islower() for char in pwd):
        taille_alphabet = taille_alphabet + 26
    if contient_majuscule(pwd):
        taille_alphabet = taille_alphabet + 26    
    if contient_chiffre(pwd):
        taille_alphabet  = taille_alphabet + 10     
    if contient_caractere_special(pwd):
        taille_alphabet  = taille_alphabet + 32
    combinaisons = taille_alphabet ** len(pwd)
    tentatives_moyennes = combinaisons // 2
    temps = tentatives_moyennes / tentatives_par_seconde
    return temps
    
    raise NotImplementedError
```

### `valider`

Retourne un **tuple** `(ok, raisons)` où :
- `ok` est `True` si toutes les règles sont satisfaites ;
- `raisons` est la liste des règles **non** satisfaites (chaînes), dans l'ordre :
  - `"trop court"`
  - `"pas de majuscule"`
  - `"pas de chiffre"`
  - `"pas de caractère spécial"`

Si `ok == True`, alors `raisons == []`.

```python
def valider(pwd: str, min_len: int = 8) -> tuple[bool, list[str]]:
    """Retourne (ok, raisons).

    raisons est la liste des règles NON satisfaites, dans l'ordre :
        "trop court", "pas de majuscule", "pas de chiffre", "pas de caractère spécial"
    """
    # TODO
    mdp = []
    raisons = []
    ok = True
    if not est_assez_long(pwd,min_len):
        raisons.append("trop court")
        ok = False
    if not contient_majuscule(pwd):
        raisons.append("pas de majuscule")
        ok = False   
    if not contient_chiffre(pwd):
        raisons.append("pas de chiffre")
        ok = False   
    if not contient_caractere_special(pwd):
        raisons.append("pas de caractère spécial")
        ok = False
    mdp.append(ok)
    mdp.append(raisons)
    return mpd
    raise NotImplementedError
```

```python
def _test_default_8():
    assert est_assez_long("abcdefgh") is True
    assert est_assez_long("abcdefg") is False
_test_default_8()

def _test_min_len_personnalise():
    assert est_assez_long("abc", min_len=3) is True
    assert est_assez_long("ab", min_len=3) is False
_test_min_len_personnalise()

def _test_vide():
    assert est_assez_long("") is False
_test_vide()
print("✅ Est Assez Long : OK")
```

```python
def _test_aucune_regle():
    assert score_force("abcde") == 0
_test_aucune_regle()

def _test_une_regle_longueur():
    assert score_force("abcdefgh") == 1
_test_une_regle_longueur()

def _test_deux_regles():
    assert score_force("Abcdefgh") == 2
_test_deux_regles()

def _test_trois_regles():
    assert score_force("Abcdefg1") == 3
_test_trois_regles()

def _test_quatre_regles():
    assert score_force("Abcdef1!") == 4
_test_quatre_regles()

def _test_court_mais_complet():
    assert score_force("Ab1!") == 3
_test_court_mais_complet()
print("✅ Score Force : OK")
```

### Temps Cassage Estime

```python
def _test_vide():
    assert temps_cassage_estime("") == 0.0
_test_vide()

def _test_minuscules_seulement():
    assert temps_cassage_estime("abc") == pytest.approx(17576 / 2 / 1e9)
_test_minuscules_seulement()

def _test_minuscule_et_chiffre():
    assert temps_cassage_estime("ab12") == pytest.approx(36**4 / 2 / 1e9)
_test_minuscule_et_chiffre()

def _test_quatre_classes():
    attendu = 94**8 / 2 / 1e9
    assert temps_cassage_estime("Ab1!cdef") == pytest.approx(attendu)
_test_quatre_classes()

def _test_tentatives_par_seconde_personnalise():
    assert temps_cassage_estime("abc", tentatives_par_seconde=100) == pytest.approx(17576 / 2 / 100)
_test_tentatives_par_seconde_personnalise()

def _test_seulement_majuscules():
    assert temps_cassage_estime("ABCD") == pytest.approx(26**4 / 2 / 1e9)
_test_seulement_majuscules()
print("✅ Temps Cassage Estime : OK")
```

### Valider

```python
def _test_mot_de_passe_valide():
    ok, raisons = valider("Abcdef1!")
    assert ok is True
    assert raisons == []
_test_mot_de_passe_valide()

def _test_trop_court():
    ok, raisons = valider("Ab1!")
    assert ok is False
    assert "trop court" in raisons
_test_trop_court()

def _test_sans_majuscule():
    ok, raisons = valider("abcdef1!")
    assert ok is False
    assert "pas de majuscule" in raisons
_test_sans_majuscule()

def _test_sans_chiffre():
    ok, raisons = valider("Abcdefg!")
    assert ok is False
    assert "pas de chiffre" in raisons
_test_sans_chiffre()

def _test_sans_special():
    ok, raisons = valider("Abcdefg1")
    assert ok is False
    assert "pas de caractère spécial" in raisons
_test_sans_special()

def _test_toutes_raisons_dans_l_ordre():
    ok, raisons = valider("abcd")
    assert ok is False
    assert raisons == [
        "trop court",
        "pas de majuscule",
        "pas de chiffre",
        "pas de caractère spécial",
    ]
_test_toutes_raisons_dans_l_ordre()

def _test_min_len_personnalise():
    ok, raisons = valider("Ab1!", min_len=4)
    assert ok is True
    assert raisons == []
_test_min_len_personnalise()
print("✅ Valider : OK")
```
