### `moyenne_eleve`

Retourne la moyenne **générale** d'un élève (toutes matières confondues, chaque note pesant
le même poids).

- Si l'élève n'a aucune note, retourne `0.0`.
- Astuce : itère sur `eleve["notes"].values()`, accumule somme et nombre.

````python
def moyenne_eleve(eleve: dict) -> float:
    """Moyenne générale d'un élève, toutes matières et notes confondues.

    Si l'élève n'a aucune note, retourne 0.0.
    """
    # TODO
    somme = 0
    cpt = 0
    #all_notes est une liste de listes donc i est une liste
    for i in eleve["notes"].values():
        cpt = cpt + len(i)
        somme = somme + sum(i)
    if cpt == 0:
        return 0.0
    else:
        return somme / cpt
    raise NotImplementedError
````

### `moyenne_par_matiere`

Retourne un **dict** `{matière: moyenne de la classe sur cette matière}`.

- Itère sur tous les élèves, puis sur leurs matières.
- Stocke pour chaque matière un total et un compteur, puis divise.
- Ignore les matières sans aucune note.

````python
def moyenne_par_matiere(classe: list[dict]) -> dict[str, float]:
    """Pour chaque matière de la classe, moyenne de toutes les notes
    de tous les élèves dans cette matière.

    Retourne un dict {matiere: moyenne}.
    Les matières qui n'ont aucune note sont ignorées.
    """
    # TODO
    total = {}
    compteur = {}
    for eleve in classe:
        for matiere,notes in eleve["notes"].items():
            if notes:
                total [matiere] = total.get(matiere,0) + sum(notes)
                compteur [matiere] = compteur.get(matiere,0) + len(notes)
    return {m: total[m]/compteur[m] for m in total}
    raise NotImplementedError
````

### `mention`

Retourne la mention française classique selon la moyenne :

| Moyenne          | Mention             |
|------------------|---------------------|
| ≥ 16             | `"Très bien"`       |
| ≥ 14 et < 16     | `"Bien"`            |
| ≥ 12 et < 14     | `"Assez bien"`      |
| ≥ 10 et < 12     | `"Passable"`        |
| < 10             | `"Insuffisant"`     |

````python
def mention(moyenne: float) -> str:
    """Retourne la mention française :
        >= 16            -> "Très bien"
        >= 14 et < 16    -> "Bien"
        >= 12 et < 14    -> "Assez bien"
        >= 10 et < 12    -> "Passable"
        < 10             -> "Insuffisant"
    """
    # TODO
    if moyenne < 10:
        return "Insuffisant"
    elif moyenne < 12:
        return "Passable"
    elif moyenne < 14:
        return "Assez bien"
    elif moyenne < 16:
        return "Bien"
    else:
        return "Très bien"
    raise NotImplementedError
````

### `top_n_eleves`

Retourne la liste des **noms** des `n` meilleurs élèves, classés par moyenne **décroissante**.

- Si `n` est plus grand que la classe, retourne tout le monde.
- En cas d'égalité, l'ordre alphabétique départage (sécurité pour les tests).
- Tu peux utiliser `sorted(...)` avec une `key`.

````python
def top_n_eleves(classe: list[dict], n: int) -> list[str]:
    """Retourne les noms des n meilleurs élèves, classés par moyenne décroissante.
    En cas d'égalité, l'ordre alphabétique départage.
    Si n est supérieur à la taille de la classe, retourne toute la classe.
    """
    # TODO
    eleve_moyenne = {}
    for x in classe:
        m = moyenne_eleve(x)
        eleve_moyenne[x["nom"]] = m

    nom_tries = sorted(eleve_moyenne.keys(),key=lambda nom: (-eleve_moyenne[nom], nom))
    
    return list(nom_tries)[:n]
    raise NotImplementedError
````

### `bulletin_classe`

Retourne un **dict** récapitulatif de la classe :

```python
{
    "nb_eleves":          int,            # nombre d'élèves
    "moyenne_generale":   float,          # moyenne des moyennes des élèves
    "meilleur_eleve":     str,            # nom de l'élève avec la meilleure moyenne
    "moins_bon_eleve":    str,            # nom de l'élève avec la moins bonne moyenne
    "moyennes_matieres":  dict[str, float],
    "mentions":           dict[str, int], # combien d'élèves ont chaque mention
}
```

C'est l'**orchestration** finale qui réutilise les fonctions précédentes.

````python
def bulletin_classe(classe: list[dict]) -> dict:
    """Synthèse de la classe :
        - nb_eleves         : nombre d'élèves
        - moyenne_generale  : moyenne des moyennes des élèves
        - meilleur_eleve    : nom de l'élève avec la meilleure moyenne
        - moins_bon_eleve   : nom de l'élève avec la moins bonne moyenne
        - moyennes_matieres : dict {matiere: moyenne de la classe}
        - mentions          : dict {mention: nb d'élèves avec cette mention}
    """
    # TODO
    bulletin ={}
    bulletin["nb_eleves"] = len(classe)
    s = 0
    for eleve in classe:
        s = s + moyenne_eleve(eleve)
    bulletin["moyenne_generale"] = s/len(classe)
    eleves = top_n_eleves(classe, len(classe))
    bulletin["meilleur_eleve"] = eleves [0]
    bulletin["moins_bon_eleve"] = eleves [len(classe)-1]
    bulletin["moyennes_matieres"] = moyenne_par_matiere(classe)

    bulletin["mentions"]  = {}

    for eleve in classe:
        m = moyenne_eleve(eleve)
        men = mention(m)
        bulletin["mentions"][men] = bulletin["mentions"].get(men,0)  +1
    return bulletin
    raise NotImplementedError
````

## Tests

### Moyenne Eleve

````python
def _test_alice():
    assert moyenne_eleve(alice) == pytest.approx(13.625)
_test_alice()

def _test_bob():
    assert moyenne_eleve(bob) == pytest.approx(11.25)
_test_bob()

def _test_eleve_sans_notes():
    eleve = {"nom": "Vide", "notes": {}}
    assert moyenne_eleve(eleve) == 0.0
_test_eleve_sans_notes()

def _test_eleve_avec_matieres_vides():
    eleve = {"nom": "X", "notes": {"maths": [], "francais": []}}
    assert moyenne_eleve(eleve) == 0.0
_test_eleve_avec_matieres_vides()
print("✅ Moyenne Eleve : OK")
````

### Moyenne Par Matiere

````python
def _test_classe_complete():
    result = moyenne_par_matiere(classe)
    # maths: (12+15+18+8+9+11+18+19+20) / 9 = 130 / 9 = 14.444...
    # francais: (10+14+12+13+15+17+18) / 7 = 99 / 7 = 14.142857...
    # histoire: (13+11+16+10+12+16+18+17) / 8 = 113 / 8 = 14.125
    assert result["maths"] == pytest.approx(130 / 9)
    assert result["francais"] == pytest.approx(99 / 7)
    assert result["histoire"] == pytest.approx(113 / 8)
_test_classe_complete()

def _test_matiere_vide_ignoree():
    classe = [{"nom": "X", "notes": {"maths": [10, 12], "vide": []}}]
    result = moyenne_par_matiere(classe)
    assert "vide" not in result
    assert result["maths"] == pytest.approx(11.0)
_test_matiere_vide_ignoree()

def _test_classe_vide():
    assert moyenne_par_matiere([]) == {}
_test_classe_vide()
print("✅ Moyenne Par Matiere : OK")
````

### Mention

### Bulletin Classe

````python
def _test_structure():
    result = bulletin_classe(classe)
    assert set(result.keys()) == {
        "nb_eleves",
        "moyenne_generale",
        "meilleur_eleve",
        "moins_bon_eleve",
        "moyennes_matieres",
        "mentions",
    }
_test_structure()

def _test_nb_eleves():
    assert bulletin_classe(classe)["nb_eleves"] == 3
_test_nb_eleves()

def _test_moyenne_generale():
    assert bulletin_classe(classe)["moyenne_generale"] == pytest.approx(14.25)
_test_moyenne_generale()

def _test_meilleur_et_moins_bon():
    result = bulletin_classe(classe)
    assert result["meilleur_eleve"] == "Charlie"
    assert result["moins_bon_eleve"] == "Bob"
_test_meilleur_et_moins_bon()

def _test_moyennes_matieres():
    result = bulletin_classe(classe)
    assert result["moyennes_matieres"]["maths"] == pytest.approx(130 / 9)
_test_moyennes_matieres()

def _test_mentions():
    result = bulletin_classe(classe)
    assert result["mentions"]["Très bien"] == 1
    assert result["mentions"]["Assez bien"] == 1
    assert result["mentions"]["Passable"] == 1
_test_mentions()
print("✅ Bulletin Classe : OK")
````
