### 1. Boucles `for` et `while`

Une **boucle** répète un bloc de code. En Python, on a deux formes principales :

```python
# for : on itère sur une collection (liste, tuple, set, dict, str, range...)
for x in [10, 20, 30]:
    print(x)

# range(n) génère 0, 1, ..., n-1 ; range(a, b) génère a, a+1, ..., b-1
for i in range(5):           # 0, 1, 2, 3, 4
    print(i)

# while : on répète tant qu'une condition est vraie
n = 100
while n > 1:
    n = n // 2               # division entière
    print(n)
```

**`enumerate`** est très utile pour avoir l'index ET la valeur :

```python
for i, prenom in enumerate(["Alice", "Bob", "Charlie"]):
    print(f"{i}: {prenom}")
# 0: Alice
# 1: Bob
# 2: Charlie
```

### 2. Listes — la structure à tout faire

Une **liste** est une **séquence ordonnée et mutable** d'éléments (qui peuvent être de
n'importe quel type, même mélangés).

```python
notes = [12, 15, 18, 9, 14]

# Accès par index (commence à 0)
notes[0]        # 12
notes[-1]       # 14   ← index négatif = depuis la fin
notes[1:3]      # [15, 18]   ← slicing : du index 1 (inclus) au 3 (exclus)
notes[::-1]     # [14, 9, 18, 15, 12]  ← inverse la liste

# Modification (mutable)
notes.append(20)              # ajoute à la fin
notes.insert(0, 7)            # insère en position 0
notes.remove(9)               # supprime la première occurrence de 9
note_max = notes.pop()        # retire et retourne le dernier
notes[1] = 16                 # modification par index
len(notes)                    # nombre d'éléments
```

**Compréhensions de liste** (très pythoniques) :

```python
carres = [x ** 2 for x in range(10)]                # [0, 1, 4, 9, ..., 81]
pairs = [x for x in notes if x % 2 == 0]            # filtre
maj = [m.upper() for m in ["alice", "bob"]]         # ["ALICE", "BOB"]
```

### 3. Tuples — les listes immuables

Un **tuple** est comme une liste, mais on ne peut **plus le modifier après création**. Utile
pour des données qui ne doivent pas changer (coordonnées, dates, retours multiples...).

```python
point = (3, 5)         # tuple
x, y = point           # déballage (unpacking) : x=3, y=5
point[0] = 10          # ❌ TypeError : tuple immuable

# Une fonction peut retourner plusieurs valeurs via un tuple
def divmod(a, b):
    return a // b, a % b      # retourne un tuple
quotient, reste = divmod(17, 5)
```

### 4. Sets — l'unicité et les opérations ensemblistes

Un **set** est une collection **non ordonnée** d'éléments **uniques**. Idéal pour
**dédoublonner** ou faire des **tests d'appartenance rapides** (O(1) au lieu de O(n) pour une
liste).

```python
voyelles = {'a', 'e', 'i', 'o', 'u', 'y'}

'a' in voyelles            # True (très rapide)
voyelles.add('w')          # ajoute (rien ne se passe si déjà présent)
voyelles.discard('z')      # supprime (silencieux si absent)

# Opérations ensemblistes
{1, 2, 3} | {3, 4, 5}      # union : {1, 2, 3, 4, 5}
{1, 2, 3} & {3, 4, 5}      # intersection : {3}
{1, 2, 3} - {3, 4, 5}      # différence : {1, 2}

# Dédoublonner une liste
unique = list(set([1, 2, 2, 3, 1, 3]))   # [1, 2, 3] (ordre non garanti)
```

### 5. Dictionnaires — clé → valeur

Un **dictionnaire** associe des **clés** à des **valeurs**. C'est la structure reine pour
représenter des données structurées.

```python
eleve = {
    "nom": "Alice",
    "age": 20,
    "notes": [12, 15, 18],
}

eleve["nom"]                  # "Alice"
eleve["mention"] = "Bien"     # ajoute une nouvelle clé
del eleve["age"]              # supprime
"nom" in eleve                # True

# Parcours
for cle in eleve:                       # parcourt les clés
    print(cle, eleve[cle])

for cle, val in eleve.items():          # clé + valeur (le plus utile)
    print(f"{cle} = {val}")

eleve.get("age", 0)           # retourne 0 si "age" absent (au lieu de KeyError)
```

**Compréhension de dictionnaire** :

```python
notes_x_2 = {nom: note * 2 for nom, note in [("a", 10), ("b", 15)]}
# {"a": 20, "b": 30}
```

### ⚠️ Pièges classiques

1. **Modifier une liste pendant qu'on l'itère**

   ```python
   nombres = [1, 2, 3, 4, 5]
   for n in nombres:
       if n % 2 == 0:
           nombres.remove(n)   # 🐛 saute des éléments !
   # → itère plutôt sur une copie : `for n in nombres[:]:`
   # → ou utilise une compréhension : `nombres = [n for n in nombres if n % 2 != 0]`
   ```

2. **Référence partagée entre listes**

   ```python
   a = [1, 2, 3]
   b = a              # b et a pointent vers LA MÊME liste !
   b.append(4)
   print(a)           # [1, 2, 3, 4]   😱
   # Pour copier : `b = a.copy()` ou `b = a[:]` ou `b = list(a)`
   ```

3. **Argument mutable par défaut**

   ```python
   def ajouter(item, liste=[]):    # ⚠️ DANGER : la liste est créée UNE SEULE FOIS
       liste.append(item)
       return liste
   ajouter(1)        # [1]
   ajouter(2)        # [1, 2] 😱
   # → utiliser : `def ajouter(item, liste=None): liste = liste or []`
   ```

4. **Dict ordonné depuis Python 3.7+**

   Avant Python 3.7, les dicts n'étaient pas ordonnés. Maintenant, l'ordre d'insertion est
   préservé. Utile à savoir mais ne pas en abuser : pour un ordre garanti, utilise une liste
   de tuples.

5. **`in` sur une grande liste = lent**

   ```python
   if x in liste:      # O(n) — parcourt jusqu'à trouver
   if x in set:        # O(1) — accès direct via hash
   if x in dict:       # O(1) — teste sur les clés
   ```

   Pour beaucoup de tests d'appartenance, convertir en set d'abord.

### 📋 Cheatsheet

```python
# Boucles
for x in collection: ...
for i in range(n): ...
for i, x in enumerate(collection): ...
while condition: ...

# Listes
liste = [1, 2, 3]
liste.append(4) ; liste.insert(0, 0) ; liste.remove(2) ; liste.pop()
liste[i] ; liste[a:b] ; liste[::-1] ; len(liste)
[x*2 for x in liste]                        # comprehension
[x for x in liste if x > 0]                 # filter

# Tuples (immuables)
t = (1, 2, 3) ; a, b, c = t

# Sets (uniques, non ordonnés)
s = {1, 2, 3} ; s.add(4) ; s.discard(2)
s | t (union) ; s & t (inter) ; s - t (diff)

# Dictionnaires
d = {"a": 1, "b": 2}
d["c"] = 3 ; del d["a"] ; d.get("z", 0)
for k, v in d.items(): ...
{k: v*2 for k, v in d.items()}              # comprehension
```

````python
def compter_occurrences(items: list[int], valeur: int) -> int:
    """Compte le nombre d'occurrences d'une valeur dans une liste.

    Approche : Compteur incrémenté à chaque correspondance.
    Complexité : O(n) - parcours complet de la liste

    Note pédagogique : Python fournit items.count(valeur) mais on implémente
    l'algorithme pour comprendre le mécanisme de comptage.
    """
    # TODO
    cpt = 0
    for item in items:
        if ( item == valeur):
            cpt = cpt + 1
    return cpt
    raise NotImplementedError
````

## `trouver_maximum`

````python
def trouver_maximum(nums: list[int]) -> int:
    """Trouve et retourne la valeur maximale dans la liste.

    Approche : Algorithme classique de recherche du maximum.
    1. Initialiser max_val avec le premier élément
    2. Parcourir tous les éléments suivants
    3. Mettre à jour max_val si on trouve plus grand

    Complexité : O(n) - un seul parcours

    Note pédagogique :
    - On pourrait utiliser max(nums) mais on veut comprendre l'algorithme
    - Initialiser à nums[0] implique que la liste ne doit pas être vide
    - Alternative : initialiser à float('-inf') et parcourir toute la liste
    """
    # TODO
    if not nums:
        return 0
    else:
        max_val = nums[0]
        for val in nums[1:len(nums)]:
            if val > max_val:
                max_val = val
        return max_val
    raise NotImplementedError
````

````python
def _test_trouver_maximum():
    assert trouver_maximum([1, 3, 7, 2, 5]) == 7
    assert trouver_maximum([-10, -3, -1, -7]) == -1
_test_trouver_maximum()
print("✅ trouver_maximum : tous les tests passent")
````

## `calculer_moyenne`

````python
def calculer_moyenne(nums: list[int]) -> float:
    """Calcule et retourne la moyenne des valeurs.

    Approche : Somme des éléments / nombre d'éléments.
    Complexité : O(n) - parcours pour la somme

    Note pédagogique :
    - Moyenne = somme / effectif
    - Retourne un float (même si tous les éléments sont int)
    - Cas spécial : liste vide retourne 0
    """
    # TODO
    if not nums:
        return 0
    else:
        return sum(nums)/len(nums)
    raise NotImplementedError
````

````python
def _test_calculer_moyenne():
    assert calculer_moyenne([1, 2, 3, 4, 5]) == 3.0
    assert calculer_moyenne([]) == 0
_test_calculer_moyenne()
print("✅ calculer_moyenne : tous les tests passent")
````

````python
def compter_mots(phrase: str) -> int:
    """Compte le nombre de mots dans une chaîne.

    Approche : Utilisation de .split() qui découpe selon les espaces.
    Complexité : O(n) où n = longueur de la chaîne

    Note pédagogique :
    - split() sans argument découpe sur n'importe quel whitespace (espace, tab, newline)
    - Gère automatiquement les espaces multiples
    - "Hello  world" -> ["Hello", "world"] (2 mots)
    """
    # TODO
    liste = phrase.split()
    return len(liste)
    raise NotImplementedError
````

## `trouver_plus_long`

````python
def trouver_plus_long(items: list[str]) -> str:
    """Trouve et retourne le mot le plus long.

    Approche : Recherche du maximum basée sur len().
    Complexité : O(n * m) où n = nombre de mots, m = longueur moyenne

    Note pédagogique : Similaire à trouver_maximum() mais on compare les longueurs.
    En cas d'égalité, on retourne le premier trouvé.
    """
    # TODO
    max_len = 0
    max_len_item = ""
    if not items:
        return ""
    else:
        for item in items:
            if len(item) > max_len:
                max_len = len(item)
                max_len_item = item
        return max_len_item
    raise NotImplementedError
````

````python
def _test_trouver_plus_long():
    assert trouver_plus_long(["chat", "chouette", "chien"]) == "chouette"
    assert trouver_plus_long([]) == ""
_test_trouver_plus_long()
print("✅ trouver_plus_long : tous les tests passent")
````

## `compter_mots_commencant_par`

````python
def compter_mots_commencant_par(items: str, lettre: str) -> int:
    """Compte les mots commençant par une lettre donnée.

    Approche :
    1. Découper la chaîne en mots avec split()
    2. Pour chaque mot, vérifier si le premier caractère correspond
    3. Utiliser lower() pour ignorer la casse

    Complexité : O(n) où n = nombre de mots

    Note pédagogique : Attention à gérer la casse (lettre.lower()).
    """
    # TODO
    cpt = 0
    if not items:
        return 0
    else:
        item = items.split()
        for i in item:
            if (i[0] == lettre):
                cpt = cpt+1
        return cpt
    raise NotImplementedError
````

## `trouver_mot_finissant_par`

````python
def trouver_mot_finissant_par(items: str, suffixe: str) -> list[str]:
    """Trouve tous les mots qui se terminent par un suffixe donné.

    Approche :
    1. Découper en mots
    2. Filtrer ceux qui se terminent par le suffixe
    3. Utiliser .endswith() pour le test

    Complexité : O(n * m) où n = nombre de mots, m = longueur du suffixe

    Note pédagogique : Construction d'une nouvelle liste par filtrage.
    Alternative élégante : [mot for mot in items.split() if mot.endswith(suffixe)]
    """
    # TODO
    suffix_list = []
    if not items:
        return []
    else:
        item = items.split()
        for i in item:
            if (i.endswith(suffixe)):
                suffix_list.append(i)
        return suffix_list
    raise NotImplementedError
````

````python
def _test_trouver_mot_finissant_par():
    assert trouver_mot_finissant_par("Bonjour tout le monde", "r") == ["Bonjour"]
    assert trouver_mot_finissant_par("Bonjour tout le monde", "out") == ["tout"]
    assert trouver_mot_finissant_par("Bonjour tout le monde", "e") == ["le", "monde"]
    assert trouver_mot_finissant_par("Bonjour tout le monde", "r") == ["Bonjour"]
    assert trouver_mot_finissant_par("", "e") == []
_test_trouver_mot_finissant_par()
print("✅ trouver_mot_finissant_par : tous les tests passent")
````

## `somme_pairs_sets`

````python
def somme_pairs_sets(nums: set[int]) -> int:
    """Calcule la somme de tous les nombres pairs dans un set.

    Approche : Identique aux listes/tuples mais sur un set.
    Complexité : O(n)

    Note pédagogique : Les sets éliminent automatiquement les doublons,
    donc chaque valeur n'est comptée qu'une fois.
    """
    # TODO
    return sum([n for n in nums if n % 2 == 0])
    raise NotImplementedError
````

````python
def _test_somme_pairs_sets():
    assert somme_pairs_sets({1, 2, 3, 4, 5, 6}) == 12
    assert somme_pairs_sets(set()) == 0
    assert somme_pairs_sets({1, 3, 5}) == 0
_test_somme_pairs_sets()
print("✅ somme_pairs_sets : tous les tests passent")
````

## `compter_occurrences_sets`

````python
def compter_occurrences_sets(items: set[int], valeur: int) -> int:
    """Vérifie si une valeur existe dans le set (0 ou 1).

    Approche : Comme les sets n'ont pas de doublons, on retourne 0 ou 1.
    Complexité : O(1) - recherche dans un set est constante en moyenne

    Note pédagogique : C'est un cas particulier !
    - Si valeur dans le set -> retourne 1
    - Sinon -> retourne 0
    Les sets ne peuvent pas avoir de doublons par définition.
    """
    # TODO
    cpt = 0
    for item in items:
        if ( item == valeur):
            cpt = cpt + 1
    return cpt
    raise NotImplementedError
````

````python
def _test_compter_occurrences_sets():
    assert compter_occurrences_sets({1, 2, 3, 4}, 2) == 1
    assert compter_occurrences_sets(set(), 7) == 0
_test_compter_occurrences_sets()
print("✅ compter_occurrences_sets : tous les tests passent")
````

## `trouver_maximum_sets`

````python
def trouver_maximum_sets(nums: set[int]) -> int:
    """Trouve et retourne le nombre maximum d'un set.

    Approche : Même algorithme que pour les listes.
    Complexité : O(n)

    Note pédagogique : On peut utiliser max(nums) directement,
    mais on implémente l'algorithme.
    """
    # TODO
    num = list(nums)
    if not num:
        return 0
    else:
        max_val = num[0]
        for val in num[1:len(num)]:
            if val > max_val:
                max_val = val
        return max_val
    raise NotImplementedError
````

````python
def _test_trouver_maximum_sets():
    assert trouver_maximum_sets({1, 3, 7, 2, 5}) == 7
    assert trouver_maximum_sets({-10, -3, -1, -7}) == -1
_test_trouver_maximum_sets()
print("✅ trouver_maximum_sets : tous les tests passent")
````

## `compter_negatifs_sets`

````python
def compter_negatifs_sets(nums: set[int]) -> int:
    """Compte le nombre de nombres négatifs dans un set.

    Approche : Compteur avec condition n < 0.
    Complexité : O(n)

    Note pédagogique : Identique à compter_negatifs() pour les listes.
    """
    # TODO
    cpt = 0
    for i in nums:
        if i < 0:
            cpt = cpt + 1

    return cpt
    raise NotImplementedError
````

````python
def ajouter_element(d: dict, cle: str, valeur: any) -> dict:
    """Ajoute une clé et sa valeur dans un dictionnaire (mutation in-place).

    Approche : Modification directe du dictionnaire.
    Complexité : O(1)

    Note pédagogique :
    - Cette fonction MODIFIE le dictionnaire passé en paramètre
    - dict[cle] = valeur ajoute ou remplace la valeur
    - Si la clé existe déjà, la valeur est remplacée
    """
    # TODO
    d[cle] = valeur
    return d
    raise NotImplementedError
````

## `fusionner_dictionnaires`

````python
def fusionner_dictionnaires(d1: dict, d2: dict) -> dict:
    """Fusionne deux dictionnaires (d2 écrase d1 en cas de conflit).

    Approche : Copie de d1 puis mise à jour avec d2.
    Complexité : O(n + m) où n, m = tailles des dicts

    Note pédagogique :
    - .update() ajoute/remplace les clés de d2 dans d1
    - Les valeurs de d2 ont la priorité en cas de conflit
    - Python 3.9+ permet : d1 | d2
    """
    # TODO
    return d1| d2
    raise NotImplementedError
````

## `inverser_dictionnaire`

````python
def inverser_dictionnaire(d: dict) -> dict:
    """Inverse un dictionnaire, échangeant les clés et les valeurs.

    Approche : Construire un nouveau dict avec valeur -> clé.
    Complexité : O(n)

    Note pédagogique :
    - ATTENTION : Si plusieurs clés ont la même valeur, seule la dernière sera conservée
    - Exemple : {a: 1, b: 1} -> {1: b} (la clé 'a' est perdue)
    - Pour gérer les doublons : utiliser dict[valeur] = liste de clés
    """
    # TODO
    d2 = {}
    for i in d.keys():
        if d[i] not in d2:
            d2[d[i]] = i
        else:
            d2[d[i]].append(i)
    return d2
    raise NotImplementedError
````

````python
def _test_inverser_dictionnaire():
    assert inverser_dictionnaire({"a": 1, "b": 2}) == {1: "a", 2: "b"}
_test_inverser_dictionnaire()
print("✅ inverser_dictionnaire : tous les tests passent")
````

## `trouver_cle_par_valeur`

````python
def trouver_cle_par_valeur(d: dict, valeur: any) -> list[str]:
    """Retourne la liste des clés correspondant à une valeur donnée.

    Approche : Parcours des paires clé-valeur en collectant toutes les correspondances.
    Complexité : O(n)

    Note pédagogique :
    - Retourne TOUTES les clés ayant cette valeur (un dict peut avoir plusieurs clés
      pointant vers la même valeur).
    - Retourne une liste vide si aucune clé ne correspond.
    """
    # TODO
    cles = []
    for i in d:
        if d[i] == valeur:
            cles.append(i)
    return cles
    raise NotImplementedError
````

## `verifier_cle_existe`

````python
def verifier_cle_existe(d: dict, cle: str) -> bool:
    """Vérifie si une clé existe dans le dictionnaire.

    Approche : Utilisation de l'opérateur 'in'.
    Complexité : O(1) en moyenne (table de hachage)

    Note pédagogique :
    - 'cle in d' est l'idiome Python recommandé
    - Alternative : d.get(cle) is not None (mais None peut être une valeur valide)
    """
    # TODO
    for i in d:
        if i == cle:
            return True
    return False
    raise NotImplementedError
````

````python
def _test_verifier_cle_existe():
    assert verifier_cle_existe({"a": 1, "b": 2}, "a") == True
    assert verifier_cle_existe({"a": 1, "b": 2}, "c") == False
_test_verifier_cle_existe()
print("✅ verifier_cle_existe : tous les tests passent")
````
