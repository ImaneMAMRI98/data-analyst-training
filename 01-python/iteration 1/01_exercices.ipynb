### 1. Variables et types de base

Une **variable** est un nom auquel on associe une valeur. En Python, on n'a **pas besoin de
déclarer le type** : il est déduit automatiquement de la valeur (typage dynamique).

```python
age = 25            # int    (entier)
prix = 19.99        # float  (nombre à virgule)
nom = "Alice"       # str    (chaîne de caractères)
est_admin = True    # bool   (True / False)
rien = None         # NoneType (absence de valeur)
```

Les types se **convertissent** explicitement avec les fonctions `int()`, `float()`, `str()`,
`bool()` :

```python
prix_str = str(19.99)         # "19.99"
n = int("42")                 # 42
moyenne = (10 + 20) / 2       # 15.0  ← la division `/` retourne TOUJOURS un float
quotient = 17 // 5            # 3     ← `//` est la division entière
reste = 17 % 5                # 2     ← `%` est le modulo (reste de la division)
puissance = 2 ** 10           # 1024  ← `**` est l'exposant
```

> 💡 **À retenir** : `//` et `%` sont **fondamentaux en algorithmique**. On les utilise pour
> tester la divisibilité, extraire des chiffres, faire des conversions de bases, etc.

### 2. Conditions

Une condition retourne un booléen (`True` ou `False`). On l'utilise avec `if / elif / else`
pour exécuter du code seulement si une condition est satisfaite.

```python
note = 14

if note >= 16:
    mention = "Très bien"
elif note >= 14:
    mention = "Bien"
elif note >= 10:
    mention = "Passable"
else:
    mention = "Insuffisant"
```

> ⚠️ **L'ordre des conditions compte** : Python évalue les `if/elif` **dans l'ordre** et s'arrête
> à la première qui est `True`. Ici, si on avait mis `if note >= 10` en premier, on n'arriverait
> jamais sur `note >= 14`.

### 3. Opérateurs de comparaison et logiques

| Opérateur | Sens                   | Exemple                    |
|-----------|------------------------|----------------------------|
| `==`      | égal à                 | `a == b`                   |
| `!=`      | différent de           | `a != b`                   |
| `<` `>`   | inférieur / supérieur  | `a < b`                    |
| `<=` `>=` | inférieur ou égal      | `a <= b`                   |
| `and`     | ET logique             | `a > 0 and a < 10`         |
| `or`      | OU logique             | `nom == "X" or nom == "Y"` |
| `not`     | NON logique            | `not est_admin`            |
| `in`      | appartient à           | `lettre in "aeiou"`        |

Python permet aussi les **comparaisons chaînées**, très lisibles :

```python
if 0 < x < 100:           # équivalent à : x > 0 and x < 100
    print("dans l'intervalle")
```

### 4. Validation d'entrée et exceptions

Une fonction **robuste** vérifie ses entrées avant de calculer. Si l'entrée est invalide, on
**lève une exception** avec `raise` :

```python
def racine_carree(x):
    if x < 0:
        raise ValueError("Pas de racine carrée d'un nombre négatif")
    return x ** 0.5
```

Les exceptions courantes :

- `ValueError` : valeur invalide (négatif, hors plage, format incorrect)
- `TypeError` : mauvais type (str au lieu d'int, etc.)
- `ZeroDivisionError` : division par zéro
- `IndexError` / `KeyError` : index/clé inexistant·e

### ⚠️ Pièges classiques

1. **`==` vs `is`** : `==` compare les **valeurs**, `is` compare les **identités d'objet**.
   Toujours `==` pour comparer des nombres ou chaînes ; `is` uniquement pour `None`
   (`if x is None`).
2. **Division `/` vs `//`** : `5 / 2` vaut `2.5` (float), `5 // 2` vaut `2` (int). Pour
   l'algorithmique, `//` est presque toujours ce qu'on veut.
3. **Conditions oubliées** : si tu écris `if a == 1 or 2:`, Python évalue
   `(a == 1) or (2)` — et `2` est toujours truthy. Bonne forme : `if a == 1 or a == 2:`,
   ou mieux : `if a in (1, 2):`.
4. **Booléens implicites** : `if liste:` est `True` si la liste est non vide. Idem pour les
   chaînes, dicts, sets. Plus pythonique que `if len(liste) > 0:`.
5. **Modifier `int` immutable** : `n = 5; n += 1` crée un **nouvel objet** ; `n` ne pointe
   plus vers le même `5`. Important pour comprendre la suite (mutation des listes en ex02).

## `est_pair`

````python
def est_pair(n: int) -> bool:
    """Vrai si le nombre est pair.

    Approche : Utilisation de l'opérateur modulo (%) pour tester la divisibilité par 2.
    Un nombre est pair si le reste de sa division par 2 est 0.

    Complexité : O(1) - opération constante

    Note pédagogique : C'est un exemple classique d'utilisation du modulo.
    On pourrait aussi faire n & 1 == 0 (opération bit à bit), mais c'est moins lisible.
    """
    # TODO
    if ( n % 2 == 0):
        return True
    else:
        return False
    raise NotImplementedError
````

````python
def _test_est_pair():
    assert est_pair(4) is True
    assert est_pair(3) is False
    assert est_pair(0) is True
_test_est_pair()
print("✅ est_pair : tous les tests passent")
````

## `est_voyelle`

````python
def est_voyelle(lettre: str) -> bool:
    """Vrai si la lettre est une voyelle.

    Approche : On vérifie si la lettre (en minuscule) appartient à l'ensemble des voyelles.
    On utilise .lower() pour gérer les majuscules et minuscules.

    Complexité : O(1) - la recherche dans un petit set est constante

    Note pédagogique : On pourrait aussi faire lettre.lower() in 'aeiouy' (string),
    mais un set est légèrement plus idiomatique Python pour un test d'appartenance.
    La conversion en minuscule évite de doubler la logique pour A-Z.
    """
    # TODO
    voyelle = 'aeiouy' 
    if lettre.lower() in voyelle:
        return True
    else:
        return False
    raise NotImplementedError
````

````python
def _test_est_voyelle():
    assert est_voyelle('a') is True
    assert est_voyelle('b') is False
    assert est_voyelle('E') is True
    assert est_voyelle('z') is False
_test_est_voyelle()
print("✅ est_voyelle : tous les tests passent")
````

## `calcul_reduction`

````python
def calcul_reduction(prix: float, taux: float) -> float:
    """Retourne le prix après remise (taux en pourcentage).

    Approche : Calcul du nouveau prix = prix - (prix * taux / 100)
    On peut aussi écrire : prix * (1 - taux / 100)

    Complexité : O(1) - opération arithmétique simple

    Note pédagogique : Attention aux validations !
    - Le taux doit être entre 0 et 100
    - Le prix devrait être positif
    - On lève une ValueError si le prix est négatif

    Exemple :
        calcul_reduction(100, 20) -> 80.0  (20% de réduction sur 100€)
    """
    # TODO
    if prix < 0:
        raise ValueError
    elif ( taux < 0 or taux > 100):
        return 0
    else:
        return prix * (1 - taux / 100)
    raise NotImplementedError
````

````python
def _test_calcul_reduction_simples():
    assert calcul_reduction(100.0, 0.0) == 100.0
    assert calcul_reduction(100.0, 10.0) == 90.0
    assert calcul_reduction(50.0, 50.0) == 25.0
_test_calcul_reduction_simples()

def _test_calcul_reduction_bornes():
    assert calcul_reduction(0.0, 10.0) == 0.0
    assert calcul_reduction(100.0, 200.0) == 0.0  # pas négatif
    try:
        calcul_reduction(-1.0, 10.0)
        raise AssertionError('attendait ValueError')
    except ValueError:
        pass
_test_calcul_reduction_bornes()
print("✅ calcul_reduction : tous les tests passent")
````

## `est_bissextile`

````python
def est_bissextile(annee: int) -> bool:
    """Vrai si année bissextile (Grégorien).

    Règle complète :
    1. Une année est bissextile si elle est divisible par 4
    2. SAUF si elle est divisible par 100 (alors elle n'est PAS bissextile)
    3. SAUF si elle est divisible par 400 (alors elle EST bissextile)

    Approche : Tester les conditions de la plus spécifique à la plus générale.
    L'ordre des tests est crucial !

    Complexité : O(1) - 3 tests maximum

    Note pédagogique : C'est un excellent exercice de logique conditionnelle.
    Erreur fréquente : tester les conditions dans le mauvais ordre.

    Exemples :
        2000 : divisible par 400 -> bissextile ✓
        1900 : divisible par 100 mais pas 400 -> PAS bissextile ✗
        2004 : divisible par 4 mais pas 100 -> bissextile ✓
        2001 : non divisible par 4 -> PAS bissextile ✗
    """
    # TODO
    if (annee % 400 == 0):
        return True
    elif (annee % 4 == 0):
        if (annee % 100 == 0):
            return False
        else:
            return True
    else:
        return False
    raise NotImplementedError
````

````python
def _test_est_bissextile():
    assert est_bissextile(2000) is True
    assert est_bissextile(1900) is False
    assert est_bissextile(2024) is True
    assert est_bissextile(2023) is False
_test_est_bissextile()
print("✅ est_bissextile : tous les tests passent")
````

## `racine_carree`

````python
def racine_carree(x: float) -> float:
    """Retourne la racine carrée d'un nombre.

    Approche : Utilisation de l'opérateur puissance ** avec exposant 0.5
    Équivalent à math.sqrt(x) mais sans import externe.

    Complexité : O(1) - opération constante

    Note pédagogique : x ** 0.5 est équivalent à √x car x^(1/2) = √x
    Alternative : on pourrait implémenter la méthode de Newton-Raphson,
    mais ce serait over-engineering pour cet exercice.

    Validation : On lève une exception si x < 0
    (on ne peut pas calculer la racine carrée d'un nombre négatif dans ℝ).
    """
    # TODO
    if (x<0):
        raise ValueError
    else:
        return x**0.5
    raise NotImplementedError
````

````python
def _test_racine_carree():
    assert racine_carree(4) == 2
    assert racine_carree(0) == 0
    assert racine_carree(9) == 3
    try:
        racine_carree(-4)
        raise AssertionError('attendait ValueError')
    except ValueError:
        pass
_test_racine_carree()
print("✅ racine_carree : tous les tests passent")
````

## `maximum_trois`

````python
def maximum_trois(a: int, b: int, c: int) -> int:
    """Renvoie le maximum de trois valeurs sans utiliser max().

    Approche : Comparaisons successives avec if/elif/else.
    On compare d'abord a et b, puis le résultat avec c.

    Complexité : O(1) - nombre constant de comparaisons

    Note pédagogique : Plusieurs approches possibles :
    - Imbrication de if: if a > b: return a if a > c else c ...
    - Ternaire : return a if a >= b and a >= c else (b if b >= c else c)
    - Simplification : max(a, max(b, c)) mais utilise max() donc interdit

    La solution présentée est la plus lisible pour des débutants.
    """
    # TODO
    max = a
    if max < b:
        max = b
    if max < c:
        max = c
    return max
    raise NotImplementedError
````

````python
def _test_maximum_trois():
    assert maximum_trois(1, 2, 3) == 3
    assert maximum_trois(3, 2, 1) == 3
    assert maximum_trois(-5, -2, -3) == -2
_test_maximum_trois()
print("✅ maximum_trois : tous les tests passent")
````

## `factorielle`

````python
def factorielle(n: int) -> int:
    """Retourne la factorielle d'un entier.

    Approche : Itération de 1 à n en accumulant le produit.
    n! = 1 × 2 × 3 × ... × n

    Cas particulier : 0! = 1 (par convention mathématique)

    Complexité : O(n) - on itère n fois

    Note pédagogique :
    1. Validation importante : on lève une ValueError pour n < 0
       car la factorielle n'est pas définie pour les nombres négatifs.
    2. Initialisation à 1 est cruciale (élément neutre de la multiplication).
    3. range(1, n+1) car range exclut la borne supérieure.

    Alternatives :
    - Récursive : plus élégante mais risque de stack overflow pour grand n
    - math.factorial(n) : mais on veut implémenter l'algorithme

    Exemples :
        factorielle(0) -> 1
        factorielle(5) -> 120  (1*2*3*4*5)
        factorielle(-1) -> ValueError
    """
    # TODO
    s = 1
    if (n == 0):
        return 1
    elif (n<0):
        raise ValueError
    else:
        for i in range(1, n+1):
            s = s*i
        return s
    raise NotImplementedError
````

````python
def _test_factorielle():
    assert factorielle(5) == 120
    assert factorielle(0) == 1
    assert factorielle(1) == 1
    try:
        factorielle(-1)
        raise AssertionError('attendait ValueError')
    except ValueError:
        pass
_test_factorielle()
print("✅ factorielle : tous les tests passent")
````

## OPTIONNEL: `convertir_en_binaire`

````python
def convertir_en_binaire(n: int) -> str:
    """Convertit un entier en représentation binaire.

    Approche : Algorithme de divisions successives par 2.
    On récupère les restes de droite à gauche, puis on inverse.

    Complexité : O(log n) - nombre de bits dans la représentation

    Note pédagogique :
    - On pourrait utiliser bin(n) qui retourne '0b101' et faire [2:] pour enlever '0b'
    - Mais ici on implémente l'algorithme classique pour comprendre le fonctionnement

    Algorithme :
    1. Cas particulier : 0 -> '0'
    2. Tant que n > 0 :
       - Ajouter (n % 2) à la représentation binaire
       - Diviser n par 2 (division entière)
    3. Inverser la chaîne car on a construit de droite à gauche

    Validation : on lève une ValueError pour n < 0
    (la conversion binaire signée n'est pas définie ici).

    Exemples :
        5 -> 101  (5 = 1*4 + 0*2 + 1*1)
        10 -> 1010  (10 = 1*8 + 0*4 + 1*2 + 0*1)

    Trace d'exécution pour n=5 :
        5 % 2 = 1, n = 5 // 2 = 2  -> '1'
        2 % 2 = 0, n = 2 // 2 = 1  -> '01'
        1 % 2 = 1, n = 1 // 2 = 0  -> '101'
        Inversion : '101'
    """
    
    # TODO
    b = ""
    if (n < 0):
        raise ValueError
    elif (n == 0 or n == 1):
        return b + str(n)
    else:
        while (n>0):
            r = n%2
            b = b + str(r)
            n = n//2
        return b[::-1]
    raise NotImplementedError
````

````python
def _test_convertir_en_binaire():
    assert convertir_en_binaire(5) == '101'
    assert convertir_en_binaire(0) == '0'
    assert convertir_en_binaire(10) == '1010'
_test_convertir_en_binaire()
print("✅ convertir_en_binaire : tous les tests passent")
````
