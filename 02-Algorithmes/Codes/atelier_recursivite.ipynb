```python
def fact_it(n):
    res = 1
    while(n>1):
        res = res *n
        n -=1
    return res
```

```python
%time %prun fact_it(5)
```

```python
def factorial_recursive(n: int) -> int:

    if ( n<2):
        return 1
    return n * factorial_recursive(n-1)
```

```python

%time %prun factorial_recursive(5)
```

```python
def fibonacci_recursive(n: int) -> int:
    if ( n < 2):
        return n
        
    return (fibonacci_recursive(n-1) + fibonacci_recursive(n-2))
```

```python
def fibonacci(n: int) -> int:
    a, b = 0, 1
    for i in range(n):
        a, b = b, a + b
    return a
```

```python
fibonacci_recursive(5)
```

```python
fibonacci(5)
```

```python
%time %prun fibonacci_recursive(10)
```

```python
%time %prun fibonacci(10)
```

```python
def gcd_recursive(a: int, b: int) -> int:
    if b == 0:
        return a
    else:
        return gcd_recursive(b,a%b)
```

```python
%time %prun gcd_recursive(21,72)
```
